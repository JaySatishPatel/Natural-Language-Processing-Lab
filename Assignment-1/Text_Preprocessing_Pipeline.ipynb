```python
import re
import emoji
import spacy
```

```python
from nltk.tokenize import word_tokenize
from nltk.tokenize import RegexpTokenizer
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import WordNetLemmatizer
```

```python
from nltk import pos_tag
```

```python
from transformers import AutoTokenizer
```

```python
nlp = spacy.load("en_core_web_sm")

bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-multilingual-cased")
```

```python
dataset = [

    "Breaking News!! India wins the match 🇮🇳🏏 #Cricket https://news.com",

    "OMG 😂😂 This fake news is spreading everywhere!!",

    "आज मौसम बहुत अच्छा है 😊",

    "COVID-19 cases are increasing rapidly in Europe.",

    "@john Check this out!! https://abc.com #Breaking",

    "Yeh news bilkul fake hai!! 😡😡"
]
```

```python
def clean_text(text):

    # Remove URLs
    text = re.sub(r"http\S+", "", text)

    # Remove mentions
    text = re.sub(r"@\w+", "", text)

    # Remove hashtags symbol
    text = text.replace("#", "")

    # Remove emojis
    text = emoji.replace_emoji(text, replace="")

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
clean_corpus = []

for text in dataset:
    clean_text_data = clean_text(text)
    clean_corpus.append(clean_text_data)

print(clean_corpus)
```

```python
print("Whitespace Tokenization\n")

for text in clean_corpus:
    tokens = text.split()
    print(tokens)
```

```python
regex = RegexpTokenizer(r"\w+")

print("\nRule-Based Tokenization\n")

for text in clean_corpus:
    tokens = regex.tokenize(text)
    print(tokens)
```

```python
print("\nSubword Tokenization\n")

for text in clean_corpus:
    tokens = bert_tokenizer.tokenize(text)
    print(tokens)
```

```python
stop_words = set(stopwords.words("english"))

print("\nStop-word Removal\n")

for text in clean_corpus:
    tokens = word_tokenize(text)
    filtered = []

    for word in tokens:
        if word.lower() not in stop_words:
            filtered.append(word)

    print(filtered)
```

```python
porter = PorterStemmer()

print("\nPorter Stemmer\n")

for text in clean_corpus:
    words = regex.tokenize(text)
    stems = [porter.stem(word) for word in words]
    print(stems)
```

```python
snow = SnowballStemmer("english")

print("\nSnowball Stemmer\n")

for text in clean_corpus:
    words = regex.tokenize(text)
    stems = [snow.stem(word) for word in words]
    print(stems)
```

```python
lemmatizer = WordNetLemmatizer()
print("\nWordNet Lemmatization\n")

for text in clean_corpus:
    words = regex.tokenize(text)
    lemmas = [lemmatizer.lemmatize(word) for word in words]
    print(lemmas)
```

```python
print("\nNLTK POS Tagging\n")

for text in clean_corpus:
    tokens = word_tokenize(text)
    tags = pos_tag(tokens)
    print(tags)
```

```python
print("\nspaCy Processing\n")

for text in clean_corpus:
    doc = nlp(text)
    print("\nSentence:", text)

    for token in doc:
        print(
            token.text,
            "| Lemma:", token.lemma_,
            "| POS:", token.pos_
        )
```

```python
sample = clean_corpus[0]
print("Original Text:")
print(sample)

print("\n------------- NLTK ----------------")
tokens = word_tokenize(sample)
print("Tokens:", tokens)
print("POS:", pos_tag(tokens))

print("\n------------- spaCy ----------------")
doc = nlp(sample)
print("Tokens:", [token.text for token in doc])
print("Lemmas:", [token.lemma_ for token in doc])
print("POS:", [(token.text, token.pos_) for token in doc])
```
